# USER DEFINED FUNCTIONS
- A **User-Defined function** is a function created by the programmer to perform specific task.
  - **(i) Parameter** - A variable specified in the function definition.
    - Example:
    ```python
    def add(a,b):
    ```
    - Here ,`a` and`b` are **parameters**
  - **(ii) Argument** - The actual value passed to a function when it is called.
    - Example:
    ```python
    add(5, 8)
    ```
    - Here, `5` and `8` are **Arguments**

### Structure of a user_defined_function
```python
 def function_name(parameters):
        statements
        return value
```    

## FUNCTION RULES
- Rule  1:Functions begin with def.
- Rule  2:Every function must have a name.
- Rule  3:Function names should be meaningful.
- Rule  4:Parameters 0 are compulsory.
- Rule  5:Colon(:) must be used after the function definition.
- Rule  6:Indentation is mandatory.
- Rule  7:A function may or may not return to value.
- Rule  8:A function executes only when called.
- Rule  9:Parameter names oshould be meaningful.
- Rule  10:Avoid using python keywords as function names.

````python
#  MARK GRADING
def grading(mark):
    if mark>=90:
        return "A+ GRADE"
    elif mark>=80:
        return "A GRADE"
    elif mark>=60:
        return "B+ GRADE"
    elif mark>=50:
        return "B GRADE"
    elif mark>=40:
        return "C+ GRADE"
    else:
        return "failed"
````

````python
grading(90)
````
