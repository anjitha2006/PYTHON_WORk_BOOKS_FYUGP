```python
price_list=[50,80,40,30]#[p1,p2,p3,p4]
quantity_list=[2,5,6,3]#[q1,q2,q3,q4]
price_vec=np.array(price_list)
quantity_vec=np.array(quantity_list)
#taking dot product of two sectors=p1q1+p2q2+p3q3+p4q4
total_expenditure=np.dot(price_vec,quantity_vec)
print("total expenditure:",total_expenditure)
```

## Linear equations

```python
# Solving a 2-market equilibrium system:
#3P1+2P2=18 (Market 1 equilibrium)
#1P1+4P2=16 (Market 2 eQuilibrium)
#the above linear equations can be written in the following matrix form.
#AX=b
#X = A_inverse.b
# Coefficient Matrix (A)and constants vector(b)
A=np.array([[3,2],[1,4]])
b=np.array([18,16])
X=np.linalg.solve(A,b)
```

```python
P1=X[0]
P2=X[1]
print(P1,P2)
```

```python
#Method 1:Direct solver
X=np.linalg.solve(A,b)
print("Equilibrium price vectors:",X)
#Method 2:Matrix inversion (X=A^-1*b)
X_inv=np.linalg.inv(A) @ b
print(f"Equilibrium price P1:{X[0]:2f}")
print(f"Equilibrium price P2:{X[1]:2f}")
```

## Eigen Values and Eigen Vectors
