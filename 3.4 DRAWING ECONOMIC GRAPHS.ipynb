## **Demand Curve**
The demand curve shows the relationship between **Price (P) and Quantity Demanded**.According to the **Law of Demand**,other things remaining constant. When price increases,quantity demanded generally decreases. Therefore, the demand curve generally slops **Downward from left to right**.

```python
import numpy as np
import pandas as pd
P=np.linspace(0,50,100)
Qd=100-2*P
df=pd.DataFrame(zip(Qd,P))
df.columns=["Quantity_Demanded","Price"]
print(df.head(10))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Price values
P=np.linspace(0,50,100)

# Demand Function Qd= f(P)
Qd=100-2*P

# Plot demand curve
plt.plot(Qd,P,label="Demand Curve")

# Labels
plt.xlabel("Quantity Demanded")
plt.ylabel("Price")

# Title
plt.title("Demand Curve")

#Legend
plt.legend()

#Grid
plt.grid()

# Display graph
plt.show()
```

## Calculate Equilibrium Using Python

```python
import sympy as sp #sympy is a symbolic mathematics library

P = sp.symbols('P')

Qd = 100 - 2 * P
Qs = -20 + 2 * P
equilibrium_price = sp.solve(Qd - Qs, P)[0]
equilibrium_quantity = Qd.subs(P, equilibrium_price)

print("Equilibrium price : ",equilibrium_price)
print("Equilibrium Quantity :",equilibrium_quantity)
```

```python
#cobb douglus production Function
import numpy as np
import matplotlib.pyplot as plt

l=np.linspace(1,100,100)

A = 1
k = 10
alpha = 0.5
beta = 0.5

Q = A*(k**alpha)*(l**beta)

plt.plot(l,Q,label ="production_function")
plt.title("cobb-Douglus production function")
plt.legend()
plt.grid()
```

```python
# Laffer curve
import numpy as np
import matplotlib.pyplot as plt

T = np.linspace(0,1,100)

revenue = 100 * T - 100 * T**2

plt.plot(T * 100, revenue,label= "Laffer curve")

plt.xlabel("Tax Rate (%)")
plt.ylabel("Tax Revenue")

plt.legend()
plt.grid

plt.show()
```
