## What is Data Visualisation
Data visualisation is the process of representing numerical or categorical data graphically

## Why is Visualisation important in Economics?
1. **Identity trends**
2. **Compare Economic Variables**
3. **Study Relationships**
4. **Identify Unusual Observations**
5. **Communicate Economic Results**

## Python Visualisation Libraries
1. **Matplotilib**
2. **Seaborn**
3. **Plotly**
4. **Pandas**

## Visualisation using Matplotlib library

```python
# Ploting of Vectors
import numpy as np
import matplotlib.pyplot as plt
plt.figure(figsize=(15,6))
year=np.array([2019,2020,2021,2022,2023,2024,2025])
gdp=np.array([52,57,60,67,70,76,81])
unemployment=np.array([10,12,15,14,20,15,13])
plt.plot(year,gdp,marker='o',label="gdp")
plt.plot(year,unemployment,marker='o',label="unemployment")
plt.title("Time Series Plot of GDP",fontdict={'fontsize':20,
        'fontweight': 'bold',
        'color': 'purple',
        'verticalalignment': 'baseline',
        'horizontalalignment':'center'})
plt.ylabel("Gross Domestic Product")
plt.xlabel("Year")
plt.legend()
plt.show()
```


# BAR GRAPH

```python
import matplotlib.pyplot as plt
states=["Kerala","Tamilnadu","Karnataka","Maharashtra","Gujarat"]
gdp=[100,150,180,250,200]
bar=plt.bar(states,gdp)
plt.xlabel("States")
plt.ylabel("GDP")
plt.title("GDP of Selected States",fontdict= {'fontsize':20,
        'fontweight':'bold',
        'color':'orange',
        'verticalalignment': 'baseline',
        'horizontalalignment':'center'})
plt.bar_label(bar)
plt.show()
```

# HORIZONTAL BAR GRAPH

```python
import matplotlib.pyplot as plt
states=["Kerala","Tamilnadu","Karnataka","Maharashtra","Gujarat"]
gdp=[100,150,180,250,200]
barh=plt.barh(states,gdp)
plt.xlabel("States")
plt.ylabel("GDP")
plt.title("GDP of Selected States",fontdict= {'fontsize':20,
        'fontweight':'bold',
        'color':'teal',
        'verticalalignment': 'baseline',
        'horizontalalignment':'center'})
plt.bar_label(barh)
plt.show()
```

# SCATTER PLOT

```python
Income=np.array([
    20,20,20,20,
    30,30,30,30,
    40,40,40,40,40,
    50,50,50,50,50,
    60,60,60,60,
    70,70,70,70,70
])
Consumption=np.array([
    15,16,13,12,
    18,19,20,25,
    22,23,25,27,26,
    29,30,32,33,35,
    35,36,38,40,
    42,43,45,47,49
])
plt.scatter(Income,Consumption)
plt.xlabel("Income")
plt.ylabel("Consumption")
plt.title("Income and Consumption")
#plt.grid
plt.show
```

# SIMPLE PIE CHART

```python
import matplotlib.pyplot as plt
sectors=["Education","Health","Defence","Infrastructure"]
expenditure=[30,20,25,25]
plt.pie(expenditure,labels=sectors,autopct="%1.2f%%",explode=[0.1,0,0,0],shadow=True)
plt.title("Government Expenditure")
plt.show()
```

# Visualisation Using Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
data={"Year":[2021,2022,2023,2024,2025],
     "GDP":[220,240,260,285,310]}
df=pd.DataFrame(data)
print(df)
```

```python
df.plot(x="Year",y="GDP",kind="scatter") # kind= scatter,bar,line,pie
plt.title("GDP Trend")
plt.xticks(rotation=90)
plt.savefig("GDPTrend.png")
plt.show()
```

```python
df.plot(x="Year",y="GDP",kind="bar")
plt.title("GDP Trend")
plt.show()
```

```python
df.plot(x="Year",y="GDP",kind="line")
plt.title("GDP Trend")
plt.show()
```
