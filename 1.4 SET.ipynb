# SETs
A set is an **unorderd**, **mutable**,and **unindexed** collection of **unique** element in python.
- Stores only unique values.
- Duplicate values are automatically removed.
- Does not support  indexing or slicing.
- Elements can be added or removed.
- Implemented using a hash table, making searching very fast

```python
# Create a Set
numset1={1,2,3,4,5}
print(numset1)
```

```python
# Empty set
A={}
print(type(A))
B= set()
print(type(B))
```

## Characteristics of Sets
- 1. Unique Elements
- 2. Unordered
- 3. Mutable
- 4. No Indexing

```python
# Unique Elements 
set1={1,2,3,45,65,14,45,65,12,1,2}
print(set1)
```

```python
# Unorderd
set2 = {1,2,16,3,4,5,16,72,15,14}
print(set2)
```

```python
# Mutable
set2.add(100)
print(set2)
```

```python
# No Indexing
set2[1]
```

## Built-in-function

```python
# len
print(len(set1))
print(len(set2))
# max
print(max(set1))
print(max(set2))
# min
print(min(set1))
print(min(set2))
# sum
print(sum(set1))
print(sum(set2))
# sorted
print(sorted(set1))
print(sorted(set2))
# any
print(any(set1))
print(any(set2))
# all
print(all(set1))
print(all(set2))
# type
print(type(set1))
print(type(set2))
# id
print(id(set1))
print(id(set2))
```

## SET-Methods

```python
print("set1:",set1)
# 1.Add - adding a single  element to the set
set1.add(100)
print(set1)
# 2.Update() - adding multiple  elements to the set
set1.update({100,200,300,400,})
print(set1)
# 3.Remove()
set1.remove(400)
print(set1)
# 4.Discard()
set1.discard({1,2,3})
print(set1)
# 5.Pop()
set1.pop()
print(set1)
# 6.Clear()
# 7.Copy()
set3= set1.copy()
print("set3:",set3)
```

## Mathematical Set Operations

```python
A={1,2,3,4}
B={3,4,5,6}
print("set A:",A)
print("set B:",B)
#1.Union
print("Union of A and B")
print(A.union(B))
print(A|B)
#2.Intersection
print("intersection of A and B")
print(A.intersection(B))
print(A&B)
#3.Difference A-B ,B-A
print("Difference of A and B")
print(A.difference(B))
print(A-B)
print("Difference of B and A")
print(B.difference(A))
print(B-A)
#4. Symmetric DIfference
print("symmetric Difference of A and B")
print(A.symmetric_difference(B))
print(A^B)
```

## Relationship Methods

```python
# issubset
print(A.issubset(B))
# issuperset
print(A.issuperset(B))
# isdisjoint
print(A.isdisjoint(B))
```
