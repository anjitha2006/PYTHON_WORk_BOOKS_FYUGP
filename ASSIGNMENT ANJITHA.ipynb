```python
# a) Define P as a symbolic variable using SymPy and specify the demand and supply functions.
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
# (a) Demand and Supply functions
P = sp.symbols('P')
Qd = 100 - 2*P
Qs = -20 + 2*P
print("Demand function:", Qd)
print("Supply function:", Qs)
```

```python
# (b) Find the Equilibrium Price by solving:Qd=Qs
equilibrium_price = sp.solve(Qd - Qs, P)[0]
print("Equilibrium Price:", equilibrium_price)
```

```python
# (c) Calculate the Equilibrium Quantity by substituting the equilibrium price into the demand function.
equilibrium_quantity = Qd.subs(P, equilibrium_price)
print("Equilibrium Quantity:", equilibrium_quantity)
```

```python
# (d) Matrix
A = np.array([[2, 1],
              [1, 2]])
B = np.array([10, 11])
print("Matrix A:")
print(A)
print("Matrix B:")
print(B)
```

```python
# (e) Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:")
print(eigenvalues)
print("Eigenvectors:")
print(eigenvectors)
```

```python
# (f) Price range and corresponding Demand and Supply
P_values = np.arange(0, 51, 5)
Qd_values = 100 - 2*P_values
Qs_values = -20 + 2*P_values
print("Price:", P_values)
print("Demand:", Qd_values)
print("Supply:", Qs_values)
```

```python
# (g) Plot Demand and Supply curves
plt.plot(P_values, Qd_values, label="Demand")
plt.plot(P_values, Qs_values, label="Supply")
plt.scatter(30, 40, color="red", label="Equilibrium Point")
plt.xlabel("Price (P)")
plt.ylabel("Quantity (Q)")
plt.title("Demand and Supply Curves")
plt.legend()
plt.grid()
plt.show()
```

```python
# (a) Calculate saving using for-loop
households = ['A', 'B', 'C', 'D', 'E']
income = [20000, 30000, 40000, 50000, 60000]
consumption = [18000, 25000, 32000, 38000, 45000]
for i in range(len(households)):
    saving = income[i] - consumption[i]
    print(households[i], "Saving =", saving)
```

```python
# (b) Classify as Saver or Dissaver
for i in range(len(households)):
    saving = income[i] - consumption[i]
    if saving > 0:
        print(households[i], "Saver")
    else:
        print(households[i], "Dissaver")
```

```python
# (c) User-defined function
def calculate_saving(income, consumption):
    return income - consumption
for i in range(len(households)):
    saving = calculate_saving(income[i], consumption[i])
    print(households[i], "Saving =", saving)
```

```python
#d) 
class Household:
    def __init__(self, income, consumption):
        self.income = income
        self.consumption = consumption

    def calculate_saving(self):
        return self.income - self.consumption

    def display(self):
        saving = self.calculate_saving()
        print("Income:", self.income)
        print("Consumption:", self.consumption)
        print("Saving:", saving)


h1 = Household(20000, 18000)
h2 = Household(30000, 25000)
h3 = Household(40000, 32000)

h1.display()
h2.display()
h3.display()
```

```python
# (e) Display all household details using while-loop
i = 0

while i < len(households):
    h = Household(income[i], consumption[i])

    print("Household:", households[i])
    h.display()
    print()

    i += 1
```
