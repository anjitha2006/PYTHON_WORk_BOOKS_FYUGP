## DICTIONARY
A dictionary is a built-in python data structure that stores data in the form of key-value pairs
- Keys are unique identifiers.
- Values are the data associated with those keys.
- Dictionaries are mutable,dynamic,and ordered.
- Dictionaries are optimized using hash tables,making key lookup very fast (average 0(1)).

```python
# creating dictionarity in key-value pairs {key:values .. . ,}
dict1 ={"name":"Anjitha", "age":19,"course":"python"}
print(dict1)
```

```python
# updating values in the dictionary
dict1["name"]="sandhya"
# change age to 25 & change course to "SQL"
dict1["age"]="24"
dict1["course"]="SQL"
print(dict1)
```

```python
# changing keys 
dict1["full_name"] = dict1["name"]
print(dict1)
dict1.pop("name")
print(dict1)
```

## Characteristics of Dictionary
1. Stores data in key value pairs
2. Mutable - update values in the dictionary and can add new key -value pairs etc.
3. Ordered - dictionary have a definite order. If we add a new key-value pair, it will positioned at last.
4. Keys must be unique
5. Keys must be immutable

### Creating Dictionary

```python
# Using curly braces 
# Using lists
name=["anjitha","sandhya","niranjana","aswathy"]
age=[20,30,19,24]
course=["python","SQL","powerB1"]
dict2={"name":name,"age": age,"course": course}
print(dict2)
# Zip method
keys= ["name","age","course"]
value= ["keerthana",20,"powerB1"]
dict3=dict(zip(keys,value))
print(dict3)
# Using Tuple
# ((key,value),(key,value),.. . .,)
tup=(("name","anu"),("age",25),("course","Economics"))
print(dict(tup))
```

```python
# Accessing Dictionary elements
# Using Keys
print(dict1["full_name"])
print(dict1["course"])
print(dict1["age"])
#to get keys and values
# 2.using get
print(dict1.get("age"))
print(dict1.get("cource"))
print(dict1.get("full_name"))
#To get all keys
print(dict1.keys())
#To get all values in the dictionary 
print("All values in dict1:",dict1.values())
#To convert all dictionary values into list 
print(list(dict1.values()))
#3. To get all items in the dictionary
print(dict1.items())
```

```python
#Adding elements 
dict1["place"]="vagamon"
dict1["Gender"]="Female"
dict1["education"]="Graduate"
print(dict1)
```

```python
#updating elements 

#using del
print(dict1)
del dict1["education"]
print(dict1)
#using_pop
dict1.pop("Gender")
print(dict1)
#using pop(specify_key)\
dict1.pop("place")
print(dict1)
#using popitem()
dict1.popitem()
print(dict1)
#using clear
```
