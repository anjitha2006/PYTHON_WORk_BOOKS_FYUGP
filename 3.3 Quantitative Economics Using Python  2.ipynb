# Calculation of Population Mean,Variance,Standard Deviation using Numpy

```python
import numpy as np
growth=[5,2,6,1,3,2,8,1,9,2]
print("Average Growth:",np.mean(growth))
print("Variance of growth:",np.var(growth))
print("Standard Deviaton:",np.std(growth))
```

```python
## Population and Sample Variance Statistics
import statistics
data=[10,20,30]
print("Population Variance:",statistics.pvariance(data))
print("Sample Variance:",statistics.variance(data))
```

# Simulating a Die Roll

```python
import random
for i in range(10):
    result=random.randint(1,6)
    print(result)
```

```python
die_roll=[1,2,3,4,5,6]
sample=[]
for i in range(10):
    result=random.choices(die_roll)
    sample.append(result[0])
    print(result[0])
print(sample)
```

```python
# Simulation of a coin toss
import random
outcomes=["Head","Tail"]
for i in range(10):
    result=random.choice(outcomes)
    print(result)
```

# Normal Distribution using python

```python
# Random.normalvariate(mu,sigma)
import random
x=random.normalvariate(100,15)
normal_dist=[]
for i in range(1000):
    x=random.normalvariate(100,15)
    normal_dist.append(x)
print(normal_dist[0:20])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.distplot(normal_dist,bins=50)
plt.title("NORMAL DISTRIBUTIONS")
plt.ylabel("DENSITY")
plt.xlabel("X VARIABLE")
# plt.legend()
plt.show()
```

# ECONOMIC GROWTH DATA SIMULATION AND VISUALISATION

```python
# Step 1: import necessary libraries
import random
import numpy as np
import statistics
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2 : Simulation of growth data with mean 6 and s.d 1.5(normal distributions)
growth=[]
for i in range(1000):
    value=random.normalvariate(6,1.5)
    growth.append(value)
    
# Step 3 : Descriptive Statistics of growth data
print("MEAN:",np.mean(growth))
print("VARIANCE:",np.var(growth))
print("STANDARD DEVIATION:",np.std(growth))

# Step 4 : Visualize the distribution:
sns.distplot(growth,bins=30)
plt.xlabel("ECONOMIC GROWTH(%)")
plt.ylabel("FREQUENCY")
plt.title("SIMULATED ECONOMIC GROWTH DISTRIBUTION")
plt.show()
```
