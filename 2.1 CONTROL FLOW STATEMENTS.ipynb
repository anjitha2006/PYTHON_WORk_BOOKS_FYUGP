```python
#a. IF statement
# age = int (input("what's your age?"))
age = int(input(" what's your age?"))
# age = 19
if age>=18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
# to deal with multiple conditions 
marks = int(input(" what's your mark?"))
#marks = 70
if marks>=90:
    print("A+ Grade")
elif marks >=80:
    print("A Grade")
elif marks >=60:
    print("B Grade")
elif marks >=40:
    print("C Grade")
else:
    print("Failed")
```

```python
#LOOPS-FOR AND WHILE LOOPS
#1.FOR LOOPS
#range function
# range(starting,ending,interval)
# for loop structure 
# for iteration_variate_name in data_structure: 
for item in range(50,0,-2):
    print(item)
```

```python
# Multiplication table of 10
num=10
for item in range(1,11):
    print(num,"x",item,"=",num*item)
```

```python
# Factorial using for loop
# n factorial = n(n-2).....2.1
n=6
fact= 1 #1*2*3
for i in range(1,n+1):
    fact*=i
print(fact)
```

```python
# Factorial using while loop
n=6
fact=1
i=1
while i<=n:
    fact*=i
    i+=1
print(fact)
```

## Controlling statements- Pass,Continue,Break

```python
# 1. Continue- Means skip the current iteration and move 
```

```python
#continue 
dept="Economics"

for i in dept:
    if i =="m":
        continue
    print(i)
```

```python
dept ="Economics"
i=0
while i<len(dept):
    if dept[i]=="m":
        pass
    print(dept[i])
    i+=1
```

```python
dept="Economics"
i=0
while i<len(dept):
    if dept[i]=="m":
        continue
    print(dept[i])
    i=i+1
```
