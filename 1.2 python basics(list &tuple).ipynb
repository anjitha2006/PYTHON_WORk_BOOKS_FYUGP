```python
# Create list
list1=[1,2,3,4,5]
list2=["anjitha","sandhya","aswathy","niranjana"]
print("1.items in list1",list1)
print("1.items in list2",list2)
# access items
print("2.first item in list1",[0])
# negative indexing
print("3.last item in list1 using negative indexing",list1[-1])
# slicing
print("5.slice the first two items from list2",list2[:2] )
print("6.slicing the last two items from list2",list2[2:])
# step slicing
print("7.to fetch 1,2,3,4,5,etc items from the list1",list1[::2])
# reverse
print("8.reversing the position of entire items list2",list2[::-1])
# update items
print("9.replace anjitha with nandhana in list2")
list2[0]="nandhana"
print(list2)
# delete items 
print("10.delete first item nandhana from list2")
del list2[0]
print(list2)
# concatenate
print("11.list concatenation",list1+list2)
# repetition
print("12.cloning items in list1",list1*2)
# membership checking
print("13.checking element 1 in list1",1 in list1)
print("14.checking element 2 in list2",2 in list2)
# non membership checking

```

# List methods

```python
#method
list5=[1,2,3,4,5,6,20,30,"jose","arjun","arun"]
#adding a single item-append(x)
list5.append(100)
print(list5)
#adding multiple items in a list-extend([multiple elements])
list5.extend(["aravindh","niranjana","sandhya","aswathi",1000,2000,3000])
print(list5)
#adding an items in a pre-assigned position-insert(i,x),
list5.insert(2,"midhun")
print(list5)
#removing items - remove(x),
list5.remove(30)
print(list5)
#pop-out or dropping the last item from the list-pop()
list5.pop()
print(list5)
#pop-out thr i-th indexed item from the list-pop(x)
list5.pop(-2)
print(list5)
#clear all the items from the list-clear()
#to get the index of the item in the list x-index()
print("index no.of niranjana",list5.index("niranjana"))
#to count the number of items from the list x-count(x)
print("count of aswathi in list5: ",list5.count("aswathi"))
#to sort the items in the list-sort()
#sort vs sorted
#sort function sorts numeric lists 
list6=[5,2,1,5,8,16,15]
list6.sort()
print("sorted list: ",list6)
sorted(list6)
#to reverse the position in the list-reverse()
list6.reverse()
print("reverese list6:",list6)
#to copy the items in the list-copy()
list6.copy()
print("copy list6:",list6)
print("copy list5:",list5)
```
