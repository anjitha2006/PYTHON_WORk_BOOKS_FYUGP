# STRING
A string is a sequence of characters written inside quotes(" " or ' ')

```python
str1=("adam smith is the father of economics")
print(str1)
```

## Length Function

```python
len(str1)
```

## print strings

```python
print(str1)
str2="anjitha is a student"
print(str2)
```

## String Indexing and slicing

```python
print(len(str1))
print(len(str2))
```

### String Indexing

```python
print(f"str2:{str2}")
print("indexing")
print(str2[0])
print(str2[1])
print(str2[2])
print("negative indexing")
print(str2[-1])
print(str2[-2])
```
