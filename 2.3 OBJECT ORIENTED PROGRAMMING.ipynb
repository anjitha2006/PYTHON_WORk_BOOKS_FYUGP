# OBJECT ORIENTED  PROGRAMMING(OOP)
OOP is a programming approach in which a program is organizes around objects rather than functions or procedures.An object represents a real world entity that contains both data(attribute) and behaviour (methods). Instead of treating data and functions seperatly, OOP combines them into a singlr unit called an object. Python fully supports object-oriented programming and provides powerful tools for creating reusable,secure,and organized programs.

OOP Terminology:The basic concept of OOP include:
- **Class:A blueprint or a template for creating objects**

```python
class student:
    def __init__(self,name,age,subject):
        self.name=name
        self.age=age
        self.subject=subject
```

```python
student1= student("Anjitha",20,"Economics")
student2= student("Sandhya",20,"Economics")
```

```python
import math
from math import sqrt
class calculator:
    def addition(self,a,b):
        return a+b
    def substration(self,a,b):
        return a-b
    def multiplication(self,a,b):
        return a*b
    def division(self,a,b):
        return a/b
    def squareroot(self,a):
        return sqrt(a)
calc= calculator()# object or instance
    
```

```python
calc.addition(12,13)
```

```python
calc.substration(67,89)
```

```python
calc.multiplication(4,5)
```

```python
calc.division(4,9)
```

```python
calc.squareroot(6)
```

```python
calc.squareroot(25)
```
